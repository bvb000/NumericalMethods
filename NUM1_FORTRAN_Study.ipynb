# Lecture 6: Root finding 1 - Bisection method
BISECTION METHOD
* root inside [a,b], where f(a) and f(b) have opposite signs
* Initial estimate: b>a ; f(a)f(b)<0
* Compute midpoint: c=(b_a)/2
* If f(b)f(c)<=0; a=c
* If f(a)f(c)<0; b=c
* Iterate until a-b<precision

REGULA FALSI METHOD
* c=[f(b)a-f(a)b]/[f(b)-f(a)]


# Lecture 7: Root finding 2 - Newton-Raphson method, Secant method

NEWTON-RAPHSON METHOD
* Initial estimate: x_0
* y=f(x_0) and y'=f'(x_0)
* New estimate: x1=x0-y/y'
* Iterate until ABS(x1-x0)<precision

SECANT METHOD
* root inside [x0,x1]
* Construct line joining points (x0,f(x0)) (x1,f(x1))
* x2=x1-f(x1)[x1-x0]/[f(x1)-f(x0)]
* x0=x1
* x2-x1
* Iterate until ABS(x1-x0)<precision

# Lecture 9: Linear Systems - Cramer Rule

CRAMER RULE (3D)
* Equations of form: ax+by+cz=d
* make matrix D = |a b c|
* make matrix Dx = |d b c|
* make matrix Dy = |a d c|
* make matrix Dz = |a b d|
* Compute the determinants of every matrix
* x=Dx/D ; y=Dy/D ; z=Dz/D

Given:
$$a_1x+b_1y+c_1z=d_1$$
$$a_2x+b_2y+c_2z=d_2$$
$$a_3x+b_3y+c_3z=d_3$$

Form Matrices:
$$D=\begin{bmatrix}
a_1 & b_1 & c_1\\
a_2 & b_2 & c_2\\
a_3 & b_3 & c_3
\end{bmatrix}$$

$$D_x=\begin{bmatrix}
d_1 & b_1 & c_1\\
d_2 & b_2 & c_2\\
d_3 & b_3 & c_3
\end{bmatrix}$$

$$D_y=\begin{bmatrix}
a_1 & d_1 & c_1\\
a_2 & d_2 & c_2\\
a_3 & d_3 & c_3
\end{bmatrix}$$

$$D_z=\begin{bmatrix}
a_1 & b_1 & d_1\\
a_2 & b_2 & d_2\\
a_3 & b_3 & d_3
\end{bmatrix}$$

Compute determinants (Rule of Sarrus):
res=m(1,1)*m(2,2)*m(3,3)+m(2,1)*m(3,2)*m(1,3)+m(3,1)*m(1,2)*m(2,3)&-m(3,1)*m(2,2)*m(1,3)-m(1,1)*m(3,2)*m(2,3)-m(2,1)*m(1,2)*m(3,3)

The roots are:
$$x=\frac{detD_x}{detD}$$
$$y=\frac{detD_y}{detD}$$
$$y=\frac{detD_z}{detD}$$


CRAMER RULE (2D)
* Equations of the form: ax+by=c
* make matrix D = |a b|
* make matrix Dx = |c b|
* make matrix Dy = |a c|
* Compute determinants of every matrix
* x=Dx/D ; y=Dy/D

Given:
$$a_1x+b_1y=c_1$$
$$a_2x+b_2y=c_2$$

Form Matrices:
$$D=\begin{bmatrix}
a_1 & b_1\\
a_2 & b_2
\end{bmatrix}$$

$$D_x=\begin{bmatrix}
c_1 & b_1\\
c_2 & b_2
\end{bmatrix}$$

$$D_y=\begin{bmatrix}
a_1 & c_1\\
a_2 & c_2
\end{bmatrix}$$

Compute the determinants (Rule of Sarrus)
res=m(1,1)*m(2,2)*m(3,3)+m(2,1)*m(3,2)*m(1,3)+&

The roots are:
$$x=\frac{detD_x}{detD}$$
$$y=\frac{detD_y}{detD}$$

#Lecture 10: Linear Regression, Curve Fitting
For a data set with values of x and y, the curve can be approximated by:
$$y\approx f(x;\beta_1,\beta_2)=\beta_1+\beta_2x$$
$\beta_1$ and $\beta_2$ are found by applying the 2D Cramer rule to:
* a1=$N$
* b1=$\sum_ix_i$
* c1=$\sum_1y_i$
* a2=$\sum_ix_i$
* b2=$\sum_ix_i^2$
* c2=$\sum_ix_iy_i$

$S_{res}$
* Square residuals
* A type of loss function (represents error rof model wrspt the dataset).

$R^2$
* Coefficient of determination
* Describes overall quality of fit
$$R^2=1-\frac{S_{res}}{S_{tot}}$$
$$S_{tot}=\sum_i(y_i-\bar{y})^2$$
$$S_{res}=\sum_i[y_i-f(x_i)]^2$$
$$\bar{y}=\frac{1}{N}\sum_{i=1}^Ny_i$$

$\varepsilon_i$
* Error
$$\varepsilon_i=y_i-f(x_i)$$

$\sigma^2$
* Variance
$$\sigma^2=\sum_{i=1}^N\frac{\varepsilon_i^2}{N-D}$$
where $D=2$ (Dimensional row vectors)

$SE$
* Standard errors for the fit parameters
$$SE(\beta_1)=\sigma\sqrt{\frac{1}{N}+\frac{\bar{x}^2}{\sum_i(x_i-\bar{x})^2}}$$
$$SE(\beta_2)=\frac{\sigma}{\sqrt{\sum_i(x_i-\bar{x})^2}}$$

#Lecture 11: Nonlinear Regression - Damped Gauss-Newton Method

* Parameters $\beta_1$ and $\beta_2$
* Nonlinear model example: exponential decay
$$f(t;\beta_1,\beta_2)=\beta_1e^{-\beta_2t}$$
* Minimize sum of squared residuals $S_{res}$
$$S_{res}(\beta)=\sum_i[y_i-f(x_i,\beta)]^2$$
* Compute partial derivatives
$$\frac{\partial S_{res}}{\partial\beta_j}=2\sum_i[f(x_i,\beta)-y_i]\frac{\partial f(x_i,\beta)}{\partial\beta_j}$$
* Linearized problem - find $\Delta\beta$ such that:
$$\sum_i[f(x_i,\beta)-y_i]\frac{\partial f(x_i,\beta)}{\partial\beta_j}=\sum_ir_iJ_{ij}=0$$
* Solve via:
$$J^TJ\Delta\beta=j^Tr\to\Delta\beta=(J^TJ)^{-1}J^Tr$$
* Gauss-Newton Method Algorithm
- Set $\beta$ to initial guess, $\alpha$ is step size
- DO k=1,maxiter
  - Construct $r_i$ and $J_{ij}$
  - Compute $\Delta\beta$
  - IF $abs(\Delta\beta)^2=\sum_j\Delta\beta_j^2\leq\epsilon$ EXIT
  - $\beta=\beta+\alpha\Delta\beta$

$$\hat{\sigma}^2=\sum^N_{i=1}\frac{\epsilon_i^2}{N-D}=\frac{S_{res}(\hat{B})}{N-D}$$
Where $\hat{\epsilon}_i=y_i-\hat{f}(x_i)$.
$$\hat{Var}(\hat{\beta})=\hat{\sigma}^2[J(\hat{\beta})^TJ(\hat{\beta}]^{-1}$$
$$\hat{SE}(\hat{\beta}_i)=\sqrt{\hat{Var}(\hat{\beta})_{i,i}}$$


#Lecture 12: Numerical Integration

* Left Reimann Sum Algorithm
  * Input: $f(x)$; boundaries $a$ and $b$, small threshold $\epsilon$
  * Initialize $N$ ($N=32$) and compute $I_{old}=I_L(N)$
$$I_L(N)=\sum^{N-1}_{k=0}f(x_k)\cdot h$$
$$h=\frac{b-a}{N}$$
$$x_k=a+k\cdot h$$
$$k=0,...,N-1$$
  * Loop:
    * Increase $N$ ($N=2N$)
    * Compute $I_{new}$
    * If $|I_{new}-I_{old}|<\epsilon$ EXIT
    * $I_{old}-I_{new}$

* Midpoint Method
$$I_M(N)=\sum^{N-1}_{k=0}f(\frac{x_k+x_{k+1}}{2})\cdot h$$
$$k=0,...,N-1$$

* Trapezoidal Method
$$I_T(N)=\sum^{N-1}_{k=0}[f(x_k)+f(x_{k+1})]\cdot \frac{h}{2}$$
$$k=0,...,N$$

* Simpson Method
$$I_S(N)=\frac{h}{6}[f(x_0)+2\sum^{N-1}{k=1}f(x_k)+4\sum^{N-1}_{k=0}f(\frac{x_k+x_{k+1}}{2})+f(x_N)]$$
$$k=0,...,N$$

#Lecture 13: Differential Equations

* Euler Method
  * Input: $f(t,x)$, $t_0$, $x_0$, step size $\Delta t$, number of steps $N$
  * Set $x=x_0$ and $t=t_0$
  * DO i=1,N
    * $fx=f(t,x)$
    * $x=x+fx\cdot\Delta t$
    * $t=t+\Delta t$

* You can write higher order ODE as a system of first order ODEs

* Verlet Method
  * Input: acceleration function $a(x)$, $t_0$, $x_0$, $v_0$, step size $\Delta t$, final time $t_f$
  * repeat $N=\frac{t_f}{t_0}$ times:
    * $v=v+a(x)\Delta t/2$
    * $x=x+v\Delta t$
    * $v=v+a(x)\Delta t/2$

#Lecture 16: Rejection Method and Monte-Carlo Integration

* Rejection Method Integration
  * Uniform random number x in range [x_min, x_max]
  * Uniform random number r in range [0,1]
  * calculate acceptance probability $A=\frac{f(x)}{f_{max}}
  * IF $r\leq A$ THEN
    * Accept X=x
  * DO WHILE loop until N Accepted Points

* Hit or Miss Integration
  * $I=\int_a^bf(x)dx\approx A$
  * N_0=0
  * DO i=1,N
    * Rand x bet a and b
    * Rand y bet 0 and M [M is upper bound of f(x)]
    * IF y<f(x) THEN
      * N_0=N_0+1
  * $A=\frac{N_0}{N}\cdot(b-a)\cdot M$

* Crude Monte Carlo Integration
  * A=0
  * DO i=1,N
    * Generate rand X_i bet [a,b]
    * Calculate f(X_i)
    * A=A+f(X_i)
  * $A=\frac{A}{N}(b-a)$
  * Variance:
    * $\sigma^2_M=(<I^2_M>-<I_M>^2)$
    * $\sigma_M~\frac{1}{\sqrt{M}}$
