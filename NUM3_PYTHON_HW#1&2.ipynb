```bash
import xarray as xr
import cartopy.crs as ccrs
import matplotlib.pyplot as plt
#import numpy as np

# Load the dataset
ds = xr.open_dataset("../../DATA/gpcp/gpcp_v01r03_daily__area115_135_5_25_MJJASO_anom_yearmean.nc", engine="netcdf4")
print(ds)

year="2000"

while (int(year)<2022):
    # Set up the plot
    fig, ax = plt.subplots(figsize=(10, 8), subplot_kw={'projection': ccrs.PlateCarree()})
    
    # Add features to the plot
    ax.coastlines()
    
    # Add gridlines with labels
    gridlines = ax.gridlines(draw_labels=True, linestyle='--', color='gray', alpha=0.7)
    gridlines.top_labels = False
    gridlines.right_labels = False
    gridlines.left_labels = True
    gridlines.bottom_labels = True
    
    # Adjust the padding and font size of the labels
    gridlines.xlabel_style = {'size': 10, 'color': 'black'}
    gridlines.ylabel_style = {'size': 10, 'color': 'black'}
    
    ax.coastlines() #plots philippines

    subset=ds.precip.sel(time=year,method='nearest')
    pc=subset.plot.pcolormesh(ax=ax, cmap='rainbow_r', vmin=-4, vmax=4, extend='both')
    
    title=f'Mean Precipitation Anomaly in {year} (mm/day)'
    plt.title(title, fontsize=16)

    if ((int(year)-2000)<10):
        plt.savefig("frame.0000"+str(int(year)-2000)+".png")
    else:
        plt.savefig("frame.000"+str(int(year)-2000)+".png")

    plt.pause(0.1)
    
    year=str(1+int(year))
```

```bash
#Anomalous Sea Surface Temperature

import xarray as xr
import cartopy.crs as ccrs
import matplotlib.pyplot as plt
import numpy as np

# Load the dataset
ds = xr.open_dataset("../../DATA/enso/data_stream-moda_stepType-avgua_anom_meanindex.nc", engine="netcdf4")
print(ds)

data=ds["sst"]
month=ds["valid_time"]

plt.figure(figsize=(20,5))
plt.scatter(month, data, color="black")
data.plot(color="black")

y1=[0.4 for a in month]
y2=[-0.4 for a in month]

plt.plot(month, y1, ":r")
plt.plot(month, y2, ":b")

title=f'Anomalous Sea Surface Temperature (°C)'
plt.title(title, fontsize=23)
plt.xlabel("")
plt.ylabel("")

plt.savefig("SSTAnomaly.png")
```

```bash
import xarray as xr
import cartopy.crs as ccrs
import matplotlib.pyplot as plt
#import numpy as np

# Load the dataset
ds = xr.open_dataset("../../DATA/enso/data_stream-moda_stepType-avgua_anom.nc", engine="netcdf4")
print(ds)

data=ds["sst"]
month=ds["valid_time"]

for i in range(len(month)):
    # Set up the plot
    fig, ax = plt.subplots(figsize=(10, 8), subplot_kw={'projection': ccrs.PlateCarree()})
    
    # Add features to the plot
    ax.coastlines()
    
    # Add gridlines with labels
    gridlines = ax.gridlines(draw_labels=True, linestyle='--', color='gray', alpha=0.7)
    gridlines.top_labels = False
    gridlines.right_labels = False
    gridlines.left_labels = True
    gridlines.bottom_labels = True
    
    # Adjust the padding and font size of the labels
    gridlines.xlabel_style = {'size': 10, 'color': 'black'}
    gridlines.ylabel_style = {'size': 10, 'color': 'black'}
    
    ax.coastlines() #plots philippines

    subset=ds.sst.sel(valid_time=month[i],method='nearest')
    pc=xr.plot.imshow(subset, cmap='rainbow', vmin=-3, vmax=3, extend='both')
    #plt.colorbar(pc,shrink=0.3)
    
    title=f"Sea Surface Temperature Anomaly {i} months from 1996-01 (°C)"
    plt.title(title, fontsize=12)

    if (i<10):
        plt.savefig("frame.0000"+str(i)+".png")
    elif (i<100):
        plt.savefig("frame.000"+str(i)+".png")
    else:
        plt.savefig("frame.00"+str(i)+".png")

    plt.pause(0.01)
```
