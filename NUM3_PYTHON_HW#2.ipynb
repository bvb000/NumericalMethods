### How to Calculate ENSO 3.4 Index
From NCAR Climate Data Guide:
> *"**Niño 3.4 (5N-5S, 170W-120W):**  The  Niño 3.4 anomalies may be thought of as representing the average equatorial SSTs across the Pacific from about the dateline to the South American coast.  The Niño 3.4 index typically uses a 5-month running mean, and El Niño or La  Niña events are defined when the  Niño 3.4 SSTs exceed +/- 0.4C for a period of six months or more. This index was adopted as the primary index for monitoring ENSO due to its ability to capture basin-scale variability (Barnston et al. 1997)."*

Link: https://climatedataguide.ucar.edu/climate-data/nino-sst-indices-nino-12-3-34-4-oni-and-tni

```bash
ddir="/mnt/c/Users/Bea/Downloads/DATA/enso/"
fname=data_stream-moda_stepType-avgua

#presumably, the data has already been cut and merged 
```

```bash
# Examining the files

cdo -s timmean ${ddir}/${fname}.nc ${ddir}/${fname}_timmean.nc
#getting the mean precipitation for all time
#results in a 2D data set, space only (x,y)

cdo -s sub ${ddir}/${fname}.nc ${ddir}/${fname}_timmean.nc ${ddir}/${fname}_anom.nc
#subtracting the all time mean from each data set to find the anomaly
#recursive subtraction
#results in 3D data set, anomaly for all time (everyday) (x,y,t)

ncview ${ddir}/${fname}_anom.nc
```

```bash
ncview -frames ${ddir}/${fname}_anom.nc
ffmpeg -r 5 -i frame.%05d.png ENSO_1996-2022.gif
#-r 10 means framerate of 10
#-i means input
#%06d means 5 zeroes before the number 1, d is a placeholder for sequential numbers
```

```bash
#Spatial averaging

cdo -s fldmean ${ddir}/${fname}_anom.nc ${ddir}/${fname}_anom_meanindex.nc
#getting the weighted average over all space (gridded data)
#results in 1D data, just anomaly of the whole Philippines over time
#note that axis will say precipitation rather than anomaly, since metadata remains unchanged

ncview ${ddir}/${fname}_anom_meanindex.nc
#download the .ps of the image pls
#filename ENSO_1996-2022.ps
```

```bash
gs -dSAFER -dBATCH -dNOPAUSE -sDEVICE=png16m -r300 -sOutputFile=ENSO_1996-2022.png ENSO_1996-2022.ps
```

```bash
#Spatial Fraction of Positive Anomaly

# first we set to 1 all points that are with a positive anomaly
cdo -s gec,0 ${ddir}/${fname}_anom.nc ${ddir}/${fname}_anom_binary.nc
#3D data set, anomaly per year (x,y,t), except 1 if positive and 0 if negative

# and now we can add up all the 1s to see what the wet area is
cdo -s fldmean ${ddir}/${fname}_anom_binary.nc ${ddir}/${fname}_anom_index.nc
#fldmean gives a weighted average over the Earth
#the number of positive points divided by the total number of points gives the fractional of positive anomaly
#resulting data should be 1D, since we are averaging over space

ncview ${ddir}/${fname}_anom_index.nc
#download the .ps of the image pls
#filename ENSO_SpatialFraction.ps
```

```bash
gs -dSAFER -dBATCH -dNOPAUSE -sDEVICE=png16m -r300 -sOutputFile=ENSO_SpatialFraction.png ENSO_SpatialFraction.ps
```
